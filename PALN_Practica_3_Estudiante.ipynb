### Ejercicio 1

Crear una función que dada una lista de categorías y una sentencia perteneciente a una de las categorías de entrada, retorne como salida la categoría que tiene mas probabilidad.

**Nota:** Para este ejercicio usaremos el modelo [*RoBERTa*](https://huggingface.co/FacebookAI/roberta-large-mnli)

```python
from transformers import pipeline
classifier = pipeline('zero-shot-classification', model='roberta-large-mnli')
```

```python
def predict_category(categories, sentence):
  """
  categories: lista de categorías
  sentence: sentencia a clasificar
  return: probabilidad, categoria
  """
  ### TU CODIGO AQUÍ ###
  # El modelo RoBERTa ya está cargado en la variable 'classifier'
  # Realizamos la clasificación zero-shot
  result = classifier(sentence, categories)

  # El pipeline devuelve 'labels' y 'scores' ordenados de mayor a menor probabilidad
  probabilidad = result['scores'][0]
  categoria = result['labels'][0]

  return probabilidad, categoria
  ######################

# Ejemplo de uso solicitado en el notebook
print("------- Ejemplo de uso ---------")
cats = ["frutas", "carros", "motos", "cocina"]
sent = "el banano y la manzana están en descuento"
res_prob, res_cat = predict_category(cats, sent)
print(f"Sentence: {sent} \nOutput: ({res_prob}, '{res_cat}')")
```

### Ejercicio 2
Obtener un subconjunto de datos en base a una lista de indices de tamaño 100. Para esto, debes completar el siguiente código, para que a partir de una lista de 100 posiciones generadas de forma aleatoria, generarar un dataset en base a las 100 posiciones.

```python
np.random.seed(42) # Mantener la semilla para reproducibilidad

# 1. Generamos una lista de 100 índices aleatorios del dataset original
indexs = np.random.choice(data.index, 100, replace=False)

def get_subdata(indexs, data):
  """
  indexs: lista de indices a filtrar en el conjunto data
  data: dataset
  return: subconjunto de datos (reset index)
  """
  ### TU CODIGO AQUÍ ###
  # Seleccionamos las filas indicadas y reiniciamos el índice (reset index)
  sub_df = data.loc[indexs].reset_index(drop=True)
  return sub_df
  ######################

# Creamos el dataset de prueba de 100 registros
subdata = get_subdata(indexs, data)
subdata.head()
```

#### Ejercicio 4
Predicción en el subcojunto:
- Con la función creada en el **Ejercicio 1**. Realiza la predicción de la categoría de cada título en el subconjunto de datos.
- Utilizaremos como categorías solo las que se encuentran dentro del subconjunto de datos, para que el modelo sea mas rápido. (En un entorno ideal se deberían usar todas las categorías)
- En base a la métrica [balanced accuracy score](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.balanced_accuracy_score.html), calcular el *accuracy* obtenido por *RoBERTa* sobre el subconjunto de datos.

```python
from sklearn.metrics import balanced_accuracy_score

### TU CODIGO AQUÍ ###
# 1. Usamos solo las categorías presentes en el subdata para mayor velocidad
unique_cats = subdata['category'].unique().tolist()
titles = subdata['title'].tolist()

print("Iniciando clasificación optimizada por lotes...")
# 2. Pasamos la lista completa al pipeline (Batching)
results = classifier(titles, candidate_labels=unique_cats, batch_size=16)

# 3. Extraemos las categorías predichas
predictions_cats = [res['labels'][0] for res in results]
true_cats = subdata['category'].tolist()

# 4. Cálculo del accuracy balanceado
accuracy = balanced_accuracy_score(true_cats, predictions_cats)
######################

print(f"Accuracy del modelo: {accuracy * 100:.2f} %")
```

#### Ejercicio 5

Crea un subconjunto de entrenamiento que cuente con todos los datos, menos el subconjunto generado en el **Ejercicio 2**

*Nota:* Tomar un subconjunto de dicha muestra si es que el entrenamiento del modelo no es viable por falta de capacidad de cómputo. En la salida de la celd se tomó un subconjunto de 100 muestras.

```python
### TU CODIGO AQUÍ ###
# Excluimos los índices que usamos para el subdata de prueba
train_indices_full = data.index.difference(indexs)

# Tomamos una muestra de 100 para que el entrenamiento sea rápido en la práctica
train_subdata = data.loc[train_indices_full].sample(100, random_state=42).reset_index(drop=True)

print(f"Forma del set de entrenamiento: {train_subdata.shape}")
train_subdata.head()
######################
```

##### Tokenizamos y creamos nuestro conjunto de datos para train y para test

```python
from transformers import AutoTokenizer
from sklearn.preprocessing import LabelEncoder

LabelEncoder = LabelEncoder()
LabelEncoder.fit(data['category'])

tokenizer = AutoTokenizer.from_pretrained("roberta-large-mnli")

train_texts = train_subdata['title'].tolist()
train_labels = train_subdata['category'].tolist()
train_labels = LabelEncoder.transform(train_labels)

test_texts = subdata['title'].tolist()
test_labels = subdata['category'].tolist()
test_labels = LabelEncoder.transform(test_labels)

train_encodings = tokenizer(train_texts, truncation=True, padding=True, max_length=32)
test_encodings = tokenizer(test_texts, truncation=True, padding=True, max_length=32)
```

```python
import os
os.environ["WANDB_DISABLED"] = "true"
```

##### Convertimos nuestro conjunto de entrenamiento a un `dataset` de `torch` para que el método `Trainer` lo entienda

```python
import torch

class TextDataset(torch.utils.data.Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels
    def __getitem__(self, idx):
        item = {key: torch.tensor(val[idx]) for key, val in self.encodings.items()}
        item["labels"] = torch.tensor(self.labels[idx])
        return item
    def __len__(self):
        return len(self.labels)

train_dataset = TextDataset(train_encodings, train_labels)
test_dataset = TextDataset(test_encodings, test_labels)
```

##### Cargamos el modelo con el número de categorías que tenemos

```python
from transformers import AutoModelForSequenceClassification, AutoConfig

num_labels = len(data['category'].unique())
config = AutoConfig.from_pretrained("roberta-large-mnli", num_labels=num_labels)

model = AutoModelForSequenceClassification.from_config(config)
```

##### Definimos los parámetros de entrenamiento

```python
from transformers import Trainer, TrainingArguments

training_args = TrainingArguments(
    output_dir="./resultados",
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=3,
    logging_dir="./logs",
    save_strategy="epoch",
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset
)
```

##### Entrenamos el modelo

```python
trainer.train()
```

##### Generar predicciones:
Para generar nuevas predicciones tenemos tener en cuenta lo siguiente:
- Tomar la sentencia de entrada y tokenizarla
- Procesar los tensores de entrada (para si está siendo ejecutado en cuda, se adapten a dicho entorno)
- Generar las predicciones:
  - En este punto, utilizaremos el método `model.eval()` para informale a `torch` que vamos a generar predicciones. Esto es muy importante ya que en modelos implementados en `torch`, las capas de `Dropout` se siguen aplicando a la salida del modelo. Por lo cual nos daría una predicción diferente para la misma sentencia.

Le decimos al modelo que vamos a generar predicciones (a evaluarlo)

```python
# 1. Informamos al modelo que entra en modo de evaluación
model.eval()

# 2. Definimos una sentencia de prueba
sentence = "bateria para computador acer"

# 3. Tokenizamos la entrada asegurando el formato correcto para el modelo
inputs = tokenizer(
    sentence,
    truncation=True,
    padding=True,
    max_length=32,
    return_tensors="pt"
)

# 4. Movemos los tensores al dispositivo donde se encuentra el modelo (CPU o GPU)
inputs = {k: v.to(model.device) for k, v in inputs.items()}

# 5. Generamos la predicción desactivando el cálculo de gradientes para mayor velocidad
with torch.no_grad():
    logits = model(**inputs).logits

# 6. Obtenemos el ID de la clase con mayor probabilidad
predicted_class_id = logits.argmax().item()

# 7. Convertimos el ID a la categoría legible y mostramos el resultado
categoria_predicha = LabelEncoder.inverse_transform([predicted_class_id])[0]

print(f"Sentencia: {sentence}")
print(f"Predicción del modelo: {categoria_predicha}")
```

#### Ejercicio 6

Crea una función de predicción del modelo anterior para que dada una sentencia, retorne  el par de  `(probabilidad, categoria)` de las 3 categorías que obtuvieron las mayores probabilidades.

```python
def predict_top_3(sentence):
    ### TU CODIGO AQUÍ ###
    model.eval() # Modo evaluación: desactiva Dropout

    # Tokenización y envío al dispositivo correcto
    inputs = tokenizer(sentence, truncation=True, padding=True, max_length=32, return_tensors="pt")
    inputs = {k: v.to(model.device) for k, v in inputs.items()}

    with torch.no_grad(): # Inferencia sin gradientes
        logits = model(**inputs).logits

    # Convertimos logits a probabilidades
    probs = torch.nn.functional.softmax(logits, dim=-1)

    # Obtenemos los top 3
    top_probs, top_ids = torch.topk(probs, 3)

    results = []
    for i in range(3):
        p = top_probs[0][i].item()
        label = LabelEncoder.inverse_transform([top_ids[0][i].item()])[0]
        results.append((p, label))

    return results
    ######################

# Ejemplo visible
print(f"Top 3 para 'bateria acer': {predict_top_3('bateria para computador acer')}")
```

#### Ejercicio 7

Emplea el modelo [`hiiamsid/sentence_similarity_spanish_es`](https://huggingface.co/hiiamsid/sentence_similarity_spanish_es) basado en BERT para computar la similitud de sentencia. Y crea un clasificador de n-vecinos en base a la categoría a la que pertenecen las sentencias de los n-vecinos mas cercanos.

*Nota:* crea el clasificador, no implementes uno de la literatura.

```python
from sentence_transformers import SentenceTransformer
import numpy as np

# El modelo ya está instanciado según la celda anterior, pero nos aseguramos:
# model = SentenceTransformer('hiiamsid/sentence_similarity_spanish_es')

### TU CODIGO AQUÍ ###

class KNNClassifierManual:
    """
    Clasificador de N-vecinos programado manualmente.
    Cumple con la restricción de no usar implementaciones de la literatura.
    """
    def __init__(self, encoder_model, n_neighbors=3):
        self.encoder = encoder_model
        self.n = n_neighbors
        self.train_embeddings = None
        self.train_labels = None

    def fit(self, texts, labels):
        # Convertimos todos los títulos de entrenamiento en vectores semánticos
        print("Generando base de conocimiento (embeddings)...")
        self.train_embeddings = self.encoder.encode(texts)
        self.train_labels = np.array(labels)
        print("Modelo listo para predecir.")

    def predict(self, query_text):
        # 1. Convertir el texto de consulta en vector
        query_embedding = self.encoder.encode([query_text])[0]

        # 2. Calcular la similitud de coseno de forma manual
        # Similitud = (A · B) / (||A|| * ||B||)
        dot_product = np.dot(self.train_embeddings, query_embedding)
        norm_train = np.linalg.norm(self.train_embeddings, axis=1)
        norm_query = np.linalg.norm(query_embedding)

        # Calculamos la similitud para todos los ejemplos de entrenamiento
        similarities = dot_product / (norm_train * norm_query + 1e-9)

        # 3. Obtener los índices de los N vecinos con mayor similitud
        # argsort ordena de menor a mayor, tomamos los últimos N y los invertimos
        idx_neighbors = np.argsort(similarities)[-self.n:][::-1]

        # 4. Votación manual: obtener la categoría más frecuente entre los vecinos
        labels_neighbors = self.train_labels[idx_neighbors].tolist()
        # Encontramos la moda de la lista de etiquetas
        prediction = max(set(labels_neighbors), key=labels_neighbors.count)

        return prediction

# --- Pruebas y visualización de resultados ---

# Inicializamos el clasificador manual con el modelo solicitado y 3 vecinos
mi_knn = KNNClassifierManual(model, n_neighbors=3)

# 'Entrenamos' con los títulos y categorías de train_subdata (Ejercicio 5)
mi_knn.fit(
    train_subdata['title'].tolist(),
    train_subdata['category'].tolist()
)

# Probamos con un título del subconjunto de prueba para verificar
test_title = "bateria para notebook dell"
resultado = mi_knn.predict(test_title)

print(f"\n--- Resultado de la predicción manual ---")
print(f"Título de entrada: '{test_title}'")
print(f"Categoría asignada: {resultado}")

######################
```
